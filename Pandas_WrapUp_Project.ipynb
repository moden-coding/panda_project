## Section 1: Choosing your dataset (read this before you start searching)

Go to **kaggle.com/datasets** and search for a topic you care about. Before you fall in love with a dataset, check it against **every** rule below. A dataset that breaks a rule will cost you days later.

### The rules

1. **Size: check it, and don't pick anything huge.** On the dataset's page, open the **Data** tab and look at the size of the **file you'll use** (not the download size Kaggle shows at the top, which is compressed and can be much smaller). There's no hard limit, but if the file takes more than about a minute to load or makes VS Code slow, pick a smaller file or a different dataset.
2. **It loads through kagglehub without logging in.** You don’t need to download anything from the Kaggle website (that does require an account). Choose from Kaggle’s Datasets section, not Competitions, since competition data makes you join and accept rules. If loading gives you an error about logging in, credentials, or accepting terms, pick a different dataset.
3. **One table is enough.** You need one file that's a table: `.csv` is best (`.parquet`, `.xlsx` and `.json` also work).
4. **Enough to work with.** At least about **500 rows** and between **6 and 40 columns**.
5. **A target.** One column you could imagine **predicting** later: a yes/no, a category, or a number (for example *survived*, *genre*, *price*).
6. **Groups and measures.** At least **2 text columns with a handful of values** (groups, like *country* or *position*) and at least **2 number columns** (measures, like *height* or *score*).
7. **You can understand the columns.** The Kaggle page explains what the columns mean, or the names are clear.
8. **Not one we've used in class.** No Pokémon, NBA, video game sales, penguins, shot logs, Olympics, NFL plays, or Portugal schools.

### 🚩 Red flags: keep looking if...
- almost every column is text (reviews, tweets, descriptions)
- there's no column you could imagine predicting
- most columns are IDs or codes you can't decode
- the Kaggle page has no description of the columns
- the file is so large it won't load comfortably

**Find a backup.** Pick a second dataset that also follows the rules, in case your first choice doesn't work out.
