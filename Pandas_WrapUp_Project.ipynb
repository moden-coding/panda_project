## Section 1: Choosing your dataset (read this before you start searching)

Go to **kaggle.com/datasets** and search for a topic you care about. Before you fall in love with a dataset, check it against **every** rule below. A dataset that breaks a rule will cost you days later.

### The rules

1. **Size: check it, and don't pick anything huge.** On the dataset's page, open the **Data** tab and look at the size of the **file you'll use** (not the download size Kaggle shows at the top, which is compressed and can be much smaller). There's no hard limit, but if the file takes more than about a minute to load or makes VS Code slow, pick a smaller file or a different dataset.
2. **It loads through kagglehub without logging in.** You don’t need to download anything from the Kaggle website (that does require an account). Choose from Kaggle’s Datasets section, not Competitions, since competition data makes you join and accept rules. If loading gives you an error about logging in, credentials, or accepting terms, pick a different dataset.
3. **One table is enough.** You need one file that's a table: `.csv` is best (`.parquet`, `.xlsx` and `.json` also work).
4. **Enough to work with.** At least about **500 rows** and between **6 and 40 columns**.
5. **A target.** One column you could imagine **predicting** later: a yes/no, a category, or a number (for example *survived*, *genre*, *price*).
6. **Groups and measures.** At least **2 text columns with a handful of values** (groups, like *country* or *position*) and at least **2 number columns** (measures, like *height* or *score*).
7. **You can understand the columns.** The Kaggle page explains what the columns mean, or the names are clear.
8. **Not one we've used in class.** No Pokémon, NBA, video game sales, penguins, shot logs, Olympics, NFL plays, or Portugal schools.

### 🚩 Red flags: keep looking if...
- almost every column is text (reviews, tweets, descriptions)
- there's no column you could imagine predicting
- most columns are IDs or codes you can't decode
- the Kaggle page has no description of the columns
- the file is so large it won't load comfortably

**Find a backup.** Pick a second dataset that also follows the rules, in case your first choice doesn't work out.

## Question stems

This section is a **menu**: you don't write anything here. Pick a stem, fill its blanks with **your own columns**, and write the finished question in your **question plan** below. **Groups** and **measures** from your inventory fit most of the blanks. Each stem shows the Olympics lab question it came from, plus two examples from other topics.

**S1. Which ___ has the highest (or lowest) ___?**  Tools: `idxmax()` / `idxmin()`, then `.loc`
- *Olympics Q8:* Who is the oldest Olympian, and in what sport?
- Which song has the most streams? Who is the artist?
- Which car has the worst mpg? What brand is it?

**S2. Is the top ___ the only one, or is there a tie?**  Tools: a mask on the max
- *Olympics Q7:* Is Yao Ming the only athlete at 226 cm? (He had 3 rows, one per Games.)
- Is the longest song the only one that long?
- Several cars might share the best mpg. How many tie?

**S3. Which ___ has the highest average ___? Does the answer change for the total?**  Tools: `groupby` with `.mean()` and `.sum()`
- *Olympics Q10:* Which sport has the tallest athletes on average?
- Which genre's songs get the most streams on average? Is a different genre on top for total streams?
- Which brand's models sell the most on average? Which brand sold the most cars in total?

**S4. Among only the ___, which ___ is highest?**  Tools: filter, then `groupby`
- *Olympics Q13:* Among only swimming golds, which team has the most?
- Among only songs from 2020, which artist has the most streams?
- Among only electric cars, which brand has the longest average range?

**S5. Make a new column ___ from ___ and ___. Which group ranks highest on it?**  Tools: a new column, then `groupby`
- *Olympics Q14:* Make a BMI column from weight and height. Which sport has the highest average BMI?
- Make `streams_per_day` from streams and days since release. Which genre ranks highest?
- Make `price_per_hp` from price and horsepower. Which brand gives the most power for the money?

**S6. How many ___ are in each group? Should you trust the smallest group?**  Tools: `size()` or `.agg(["mean", "count"])`
- *Olympics Q14:* Tug-of-War came second in BMI, but on only 22 athletes.
- How many songs does each genre have? Is the top genre's average based on only a few songs?
- How many cars does each brand have? Can you trust a brand's average if it has only 3 cars?

**A good question** is specific (you know exactly which columns it uses), answerable with this data, and something you actually want to know. **Predict the answer before you run any code.** Being wrong is often the most interesting part.

### Weak vs. strong questions

| Weak | Strong | What makes it stronger |
|---|---|---|
| What's the average height? | Which sport has the tallest athletes on average, and is the top sport based on enough athletes to trust? | It compares groups instead of giving one number, and it checks group size. |
| How many medals are there? | Among only swimming golds, which team has the most, and how far ahead is it? | It filters, then groups, and the answer includes a comparison. |
| Do women compete in the Olympics? | How has the share of entries by women changed from 1900 to 2016? | A yes/no needs no code; a change over time is something you can measure. |

A strong question usually **compares** something, has an answer you **can't guess for sure**, and uses at least **two columns**.
